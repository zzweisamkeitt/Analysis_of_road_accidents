```python
# ============ 0) Установка (если нужно) ============
# Запусти один раз в Jupyter/Colab/VSCode Notebook
!pip -q install folium seaborn

# ============ 1) Импорты ============
import os
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

import folium
from folium.plugins import HeatMap, MarkerCluster

# Чтобы графики показывались в ноутбуке
%matplotlib inline

# ============ 2) Пути ============
CARDS_PATH = "Downloads/ДТП_карточки_21-26.csv"
PARTS_PATH = "Downloads/ДТП_участники_21-26.csv"

# (опционально) куда сохранять результаты
OUT_DIR = "./out_eda_tuva"
os.makedirs(OUT_DIR, exist_ok=True)

# ============ 3) Загрузка ============
cards = pd.read_csv(CARDS_PATH, sep=";", encoding="utf-8-sig").dropna(how="all")
parts = pd.read_csv(PARTS_PATH, sep=";", encoding="utf-8-sig").dropna(how="all")

# Часто пустые строки оставляют NaN в "Номер"
if "Номер" in cards.columns:
    cards = cards.dropna(subset=["Номер"]).copy()

# ============ 4) Типы / очистка ============
for col in ["Широта", "Долгота"]:
    if col in cards.columns:
        cards[col] = pd.to_numeric(cards[col], errors="coerce")

for col in ["Количество ТС", "Число участников", "Число погибших", "Число раненых"]:
    if col in cards.columns:
        cards[col] = pd.to_numeric(cards[col], errors="coerce").fillna(0)

# Дата/время
cards["Дата"] = pd.to_datetime(cards["Дата"], format="%d.%m.%Y", errors="coerce")
cards["Время"] = cards["Время"].fillna("00:00").astype(str)

cards["datetime"] = pd.to_datetime(
    cards["Дата"].dt.strftime("%Y-%m-%d") + " " + cards["Время"],
    errors="coerce"
)

cards["hour"] = cards["datetime"].dt.hour
cards["weekday"] = cards["datetime"].dt.day_name()

weekday_order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]

# Координаты
coords = cards[["Широта","Долгота"]].dropna().copy()
if len(coords) == 0:
    raise ValueError("После очистки не осталось координат (Широта/Долгота).")

center_lat = float(coords["Широта"].median())
center_lon = float(coords["Долгота"].median())

# Вес тяжести: раненые + 3*погибшие (коэфф. можно менять)
cards["severity_weight"] = cards["Число раненых"] + 3 * cards["Число погибших"]

# ============ 5) Быстрые таблицы (выводим прямо в ноут) ============
print("Карточки ДТП:", cards.shape)
print("Участники/ТС:", parts.shape)

display(cards.head(10))
display(cards.describe(include="all").transpose().head(30))

# Топ-адреса/дороги
if "Адрес" in cards.columns:
    display(cards["Адрес"].fillna("Не указано").value_counts().head(15).to_frame("count"))

if "Дорога" in cards.columns:
    display(cards["Дорога"].fillna("Не указано").value_counts().head(15).to_frame("count"))

# ==========================================================
# ВИЗУАЛИЗАЦИЯ 1: Интерактивная HEATMAP по координатам
# ==========================================================
m1 = folium.Map(location=[center_lat, center_lon], zoom_start=7, tiles="OpenStreetMap")
HeatMap(coords.values.tolist(), radius=18, blur=15, min_opacity=0.35).add_to(m1)
m1  # <- просто последней строкой, чтобы отрисовалось в ноутбуке

# (опционально сохранить)
m1.save(os.path.join(OUT_DIR, "01_heatmap_points.html"))

# ==========================================================
# ВИЗУАЛИЗАЦИЯ 2: “Тяжёлая” HEATMAP (вес = тяжесть ДТП)
# ==========================================================
wdf = cards.dropna(subset=["Широта","Долгота"]).copy()
if wdf["severity_weight"].sum() == 0:
    wdf["severity_weight"] = 1

m2 = folium.Map(location=[center_lat, center_lon], zoom_start=7, tiles="OpenStreetMap")
HeatMap(
    wdf[["Широта","Долгота","severity_weight"]].values.tolist(),
    radius=20, blur=18, min_opacity=0.35
).add_to(m2)
m2

m2.save(os.path.join(OUT_DIR, "02_heatmap_severity.html"))

# ==========================================================
# ВИЗУАЛИЗАЦИЯ 3: Кластер точек + popup (красиво для отчёта)
# ==========================================================
m3 = folium.Map(location=[center_lat, center_lon], zoom_start=7, tiles="OpenStreetMap")
cluster = MarkerCluster().add_to(m3)

cols_popup = [c for c in ["datetime","Вид ДТП","Состояние погоды","Освещение","Адрес","Дорога","Число погибших","Число раненых"] if c in cards.columns]
sample = cards.dropna(subset=["Широта","Долгота"]).copy()

# Чтобы карта не была “тяжёлой”, можно ограничить точки (например 1000)
MAX_POINTS = 1500
if len(sample) > MAX_POINTS:
    sample = sample.sample(MAX_POINTS, random_state=42)

for _, r in sample.iterrows():
    popup_text = "<br>".join([f"<b>{c}</b>: {r[c]}" for c in cols_popup])
    folium.Marker(
        location=[r["Широта"], r["Долгота"]],
        popup=folium.Popup(popup_text, max_width=350)
    ).add_to(cluster)

m3
m3.save(os.path.join(OUT_DIR, "03_markercluster_popups.html"))

# ==========================================================
# ВИЗУАЛИЗАЦИЯ 4: Heatmap “час × день недели”
# ==========================================================
tmp = cards.dropna(subset=["hour","weekday"]).copy()

value_col = "Номер ДТП" if "Номер ДТП" in tmp.columns else ("Номер" if "Номер" in tmp.columns else tmp.columns[0])

pivot_time = tmp.pivot_table(
    index="hour",
    columns="weekday",
    values=value_col,
    aggfunc="count",
    fill_value=0
).reindex(columns=weekday_order)

plt.figure(figsize=(10,6))
sns.heatmap(pivot_time, annot=False)
plt.title("Частота ДТП: час × день недели")
plt.xlabel("День недели")
plt.ylabel("Час")
plt.tight_layout()
plt.show()

# ==========================================================
# ВИЗУАЛИЗАЦИЯ 5: Топ видов ДТП (barh) + доли
# ==========================================================
if "Вид ДТП" in cards.columns:
    top_types = cards["Вид ДТП"].fillna("Не указано").value_counts().head(12)
    display((top_types / top_types.sum() * 100).round(1).to_frame("%"))

    plt.figure(figsize=(10,6))
    plt.barh(top_types.index[::-1], top_types.values[::-1])
    plt.title("Топ-12 видов ДТП")
    plt.xlabel("Количество")
    plt.ylabel("Вид ДТП")
    plt.tight_layout()
    plt.show()

# ==========================================================
# ВИЗУАЛИЗАЦИЯ 6: Heatmap “погода × вид ДТП” (только топ категории)
# ==========================================================
if "Вид ДТП" in cards.columns and "Состояние погоды" in cards.columns:
    df2 = cards.copy()
    df2["Вид ДТП"] = df2["Вид ДТП"].fillna("Не указано")
    df2["Состояние погоды"] = df2["Состояние погоды"].fillna("Не указано")

    top_acc = df2["Вид ДТП"].value_counts().head(10).index
    top_w = df2["Состояние погоды"].value_counts().head(8).index

    df2 = df2[df2["Вид ДТП"].isin(top_acc) & df2["Состояние погоды"].isin(top_w)]

    pivot_w = df2.pivot_table(
        index="Вид ДТП",
        columns="Состояние погоды",
        values=value_col,
        aggfunc="count",
        fill_value=0
    )

    plt.figure(figsize=(12,6))
    sns.heatmap(pivot_w, annot=False)
    plt.title("ДТП: вид × погода (топ категории)")
    plt.xlabel("Состояние погоды")
    plt.ylabel("Вид ДТП")
    plt.tight_layout()
    plt.show()

print("Готово! (если сохраняла) HTML карты лежат в:", os.path.abspath(OUT_DIR))
```

```python
import pandas as pd
import numpy as np
import folium
from folium.plugins import HeatMap, MarkerCluster, Fullscreen, MiniMap


# --- 2) Координаты ---
cards["Широта"]  = pd.to_numeric(cards["Широта"], errors="coerce")
cards["Долгота"] = pd.to_numeric(cards["Долгота"], errors="coerce")
cards = cards.dropna(subset=["Широта", "Долгота"]).copy()

# Убираем явные выбросы (на всякий случай)
cards = cards[(cards["Широта"].between(49, 56)) & (cards["Долгота"].between(88, 100))].copy()

# --- 3) Числовые поля ---
for c in ["Число погибших", "Число раненых"]:
    if c in cards.columns:
        cards[c] = pd.to_numeric(cards[c], errors="coerce").fillna(0)
    else:
        cards[c] = 0

# --- 4) Тяжесть: смертельные ДТП (есть погибшие) ---
# 10 если есть погибшие, иначе 0
cards["severity"] = np.where(cards["Число погибших"] > 0, 10, 0)

# Центр карты
center = [float(cards["Широта"].median()), float(cards["Долгота"].median())]

# --- 5) Базовая карта ---
m = folium.Map(
    location=center,
    zoom_start=7,
    tiles="CartoDB positron",
    control_scale=True
)

Fullscreen().add_to(m)
MiniMap(toggle_display=True).add_to(m)

# --- СЛОЙ 1: Плотность ДТП (все ДТП одинаково) ---
heat_layer = folium.FeatureGroup(name="Плотность ДТП (все ДТП)", show=True)
HeatMap(
    data=cards[["Широта", "Долгота"]].values.tolist(),
    radius=18,
    blur=20,
    min_opacity=0.25
).add_to(heat_layer)
heat_layer.add_to(m)

# --- СЛОЙ 2: Смертельные ДТП (есть погибшие) ---
fatal = cards[cards["severity"] > 0].copy()

weighted_layer = folium.FeatureGroup(name="Смертельные ДТП (есть погибшие)", show=False)

if len(fatal) > 0:
    w = fatal["severity"].to_numpy()  # тут будет 10
    HeatMap(
        data=np.c_[fatal["Широта"].to_numpy(), fatal["Долгота"].to_numpy(), w].tolist(),
        radius=14,        # меньше радиус — точнее локализация
        blur=16,
        min_opacity=0.35
    ).add_to(weighted_layer)
else:
    # Если вдруг нет погибших в данных
    folium.Marker(
        location=center,
        popup="В данных нет ДТП с погибшими (severity > 0)."
    ).add_to(weighted_layer)

weighted_layer.add_to(m)

# --- СЛОЙ 3: Кластеры точек + pop-up ---
cluster_layer = folium.FeatureGroup(name="Кластеры (точки + pop-up)", show=False)
cluster = MarkerCluster().add_to(cluster_layer)

# Ограничим точки для pop-up, чтобы карта не тормозила
MAX_POINTS = 1200
view = cards.copy()
if len(view) > MAX_POINTS:
    view = view.sample(MAX_POINTS, random_state=42)

popup_cols = [c for c in ["Дата","Время","Вид ДТП","Состояние погоды","Освещение","Адрес","Дорога","Число погибших","Число раненых"]
              if c in view.columns]

for _, r in view.iterrows():
    txt = "<br>".join([f"<b>{c}</b>: {r[c]}" for c in popup_cols])
    folium.CircleMarker(
        location=[r["Широта"], r["Долгота"]],
        radius=4,
        weight=0,
        fill=True,
        fill_opacity=0.6,
        popup=folium.Popup(txt, max_width=380)
    ).add_to(cluster)

cluster_layer.add_to(m)

# --- Переключатель слоёв ---
folium.LayerControl(collapsed=False).add_to(m)

# --- Показ карты в ноутбуке ---
m
```

```python
import pandas as pd
import numpy as np
import folium
from folium.plugins import HeatMap, Fullscreen

cards["Широта"]  = pd.to_numeric(cards["Широта"], errors="coerce")
cards["Долгота"] = pd.to_numeric(cards["Долгота"], errors="coerce")
cards = cards.dropna(subset=["Широта","Долгота"]).copy()

# время (для "ночь/день")
cards["Дата"] = pd.to_datetime(cards["Дата"], format="%d.%m.%Y", errors="coerce")
cards["Время"] = cards["Время"].fillna("00:00").astype(str)
cards["datetime"] = pd.to_datetime(cards["Дата"].dt.strftime("%Y-%m-%d") + " " + cards["Время"], errors="coerce")
cards["hour"] = cards["datetime"].dt.hour

center = [float(cards["Широта"].median()), float(cards["Долгота"].median())]
m = folium.Map(location=center, zoom_start=7, tiles="CartoDB positron", control_scale=True)
Fullscreen().add_to(m)

def add_heat(df, name, show=False, radius=18, blur=20):
    layer = folium.FeatureGroup(name=name, show=show)
    if len(df) > 0:
        HeatMap(df[["Широта","Долгота"]].values.tolist(), radius=radius, blur=blur, min_opacity=0.25).add_to(layer)
    layer.add_to(m)

# День/ночь
day = cards[(cards["hour"] >= 7) & (cards["hour"] <= 21)]
night = cards[(cards["hour"] < 7) | (cards["hour"] > 21)]
add_heat(cards, "Все ДТП", show=True)
add_heat(day, "ДТП днём (07–21)", show=False)
add_heat(night, "ДТП ночью (22–06)", show=False, radius=16, blur=18)

# Плохая погода (подстрой под названия в твоих данных)
if "Состояние погоды" in cards.columns:
    bad_weather_mask = cards["Состояние погоды"].astype(str).str.contains("снег|дожд|туман|метел", case=False, na=False)
    add_heat(cards[bad_weather_mask], "ДТП в плохую погоду", show=False)

# Пешеходы (подстрой под “Вид ДТП”)
if "Вид ДТП" in cards.columns:
    ped_mask = cards["Вид ДТП"].astype(str).str.contains("пешеход", case=False, na=False)
    add_heat(cards[ped_mask], "Наезды на пешеходов", show=False, radius=14, blur=16)

folium.LayerControl(collapsed=False).add_to(m)
m
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = cards.copy()
df["Дата"] = pd.to_datetime(df["Дата"], format="%d.%m.%Y", errors="coerce")
df["Время"] = df["Время"].fillna("00:00").astype(str)
df["datetime"] = pd.to_datetime(df["Дата"].dt.strftime("%Y-%m-%d") + " " + df["Время"], errors="coerce")
df = df.dropna(subset=["datetime"]).copy()

df["hour"] = df["datetime"].dt.hour

# период суток
bins = [-1, 6, 11, 16, 23]
labels = ["Ночь (0–6)", "Утро (7–11)", "День (12–16)", "Вечер (17–23)"]
df["period"] = pd.cut(df["hour"], bins=bins, labels=labels)

dow_map = {"Monday":"Пн","Tuesday":"Вт","Wednesday":"Ср","Thursday":"Чт","Friday":"Пт","Saturday":"Сб","Sunday":"Вс"}
df["dow"] = df["datetime"].dt.day_name().map(dow_map)
dow_order = ["Пн","Вт","Ср","Чт","Пт","Сб","Вс"]

pivot = df.pivot_table(index="period", columns="dow", values="Номер", aggfunc="count", fill_value=0).reindex(columns=dow_order)

plt.figure(figsize=(10,3.8))
sns.heatmap(pivot, annot=True, fmt="d", linewidths=0.3, linecolor="white")
plt.title("ДТП: период суток × день недели")
plt.xlabel("День недели")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

hour_counts = df.groupby("hour").size().reindex(range(24), fill_value=0)

plt.figure(figsize=(10,4))
plt.plot(hour_counts.index, hour_counts.values, marker="o")
plt.title("Распределение ДТП по часам (все дни вместе)")
plt.xlabel("Час")
plt.ylabel("Количество ДТП")
plt.xticks(range(0,24,1))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
dow_counts = df.groupby("dow").size().reindex(dow_order, fill_value=0)

plt.figure(figsize=(9,3.8))
plt.bar(dow_counts.index, dow_counts.values)
plt.title("Количество ДТП по дням недели")
plt.xlabel("День недели")
plt.ylabel("Количество ДТП")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = cards.copy()

col = "Вид ДТП"
s = df[col].fillna("Не указано")

share = (s.value_counts(normalize=True) * 100)
top_n = 8

top = share.head(top_n)
other = pd.Series({"Прочие": share.iloc[top_n:].sum()})
plot_data = pd.concat([top, other])

# рисуем
plt.figure(figsize=(10,6))
plt.barh(plot_data.index[::-1], plot_data.values[::-1])
plt.title("Структура ДТП по виду (доля, %)")
plt.xlabel("%")
plt.ylabel("Вид ДТП")

# подписи %
for i, v in enumerate(plot_data.values[::-1]):
    plt.text(v + 0.2, i, f"{v:.1f}%")

plt.tight_layout()
plt.show()
```
